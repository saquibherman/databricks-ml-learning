# Lecture — Managing and Exploring Data

## Overview

This lecture introduces the four platform building blocks that the rest of the course relies on:

- **A. The Databricks Data + AI Platform:** the lakehouse architecture and the workloads it supports for ML
- **B. Notebooks for ML Workflows:** exploratory data analysis, collaboration, and Genie Code
- **C. Unity Catalog:** governance for data and AI assets, the three-level namespace, permissions, and Data profiling
- **D. Delta Lake for ML:** reliable storage, time travel, the open ecosystem, and ingesting data into bronze, silver, and gold tables

## Learning Objectives

By the end of this lecture, you will be able to:

- Explain how the lakehouse architecture supports each stage of an ML workflow
- Explain how Databricks notebooks and Genie Code speed up exploratory data analysis
- Describe how Unity Catalog governs data and AI assets, and explain its three-level namespace
- Describe how Data profiling in Unity Catalog automates data-quality and drift checks on ML tables
- Explain how Delta Lake supports reliable, reproducible ML datasets, including time travel
- Describe how Delta Lake works with other engines, and choose an ingestion option for landing ML data in a medallion architecture

### C1. The Unity Catalog Architecture

- Unity Catalog sits between the **compute engine** you use to query data and the **open storage formats** your data is written in.
- Compute engines (Databricks Runtime, Spark, and others) talk to UC over the **Unity REST** and **Iceberg REST** APIs.
- UC governs data and AI assets such as tables, volumes/files, functions, and models across open formats (Delta Lake, Iceberg, Parquet), and also provides discovery and lineage across related assets such as notebooks, jobs, and dashboards.

<ul style="margin: 4px 0 0 16px;">
<li><strong>Liquid Clustering</strong> — replaces static partitioning and ZORDER. Self-tuning and skew-resistant: only rewrites the data that needs reorganization, and clustering keys can be changed at any time without rewriting the full table. Databricks recommends it for all new tables; it is generally available for Delta tables in Databricks Runtime 15.4 LTS and above. Use <code>CLUSTER BY AUTO</code> to let Databricks pick keys automatically (DBR 15.4+).</li>
<li><strong>Predictive Optimization</strong> — automatically runs OPTIMIZE, VACUUM, and ANALYZE on Unity Catalog managed tables, choosing the tables that would benefit from maintenance. No manual maintenance scheduling needed. Requires the Premium plan and UC managed tables.</li>
<li><strong>Change Data Feed</strong> — when enabled, records every row-level insert, update, and delete between table versions. Ideal for incremental ETL (process only changed rows), audit trails, and syncing changes to downstream tables or external systems. Read changes with <code>table_changes()</code> or <code>.option("readChangeFeed", True)</code>.</li>
<li><strong>MERGE</strong> — SQL upsert that matches source rows against the target on a join key, then updates matched rows and inserts new ones. Supports <code>WHEN NOT MATCHED BY SOURCE</code> for soft-deletes. MERGE does not remove duplicates from the source: if several source rows match the same target row, the MERGE fails, so deduplicate the source first.</li>
<li><strong>Schema evolution</strong> — add new columns without rewriting existing data, using <code>.option("mergeSchema", "true")</code> on write or <code>ALTER TABLE ... ADD COLUMNS</code>. Schema enforcement still rejects writes with incompatible types.</li>
<li><strong>Data skipping and scalable metadata</strong> — the transaction log stores per-file statistics (such as min and max values), so queries skip files that cannot match a filter. Metadata is handled like data, so tables with billions of files stay fast.</li>
<li><strong>Unified batch and streaming</strong> — the same Delta table can be a batch source, a streaming source, and a streaming sink, so ingestion pipelines and model-training reads share one copy of the data.</li>
<li><strong>Disk cache</strong> — Databricks caches copies of remote Parquet data on the local storage of compute nodes (formerly called the Delta cache), so repeated reads of the same training data are faster. Databricks manages it automatically; there is nothing to configure for this course.</li>

### D2. Writing a Delta Table

`saveAsTable(...)` writes Delta by default and registers the table in your current UC catalog and schema. Two write options are worth knowing:

- `mode("overwrite")` — replaces the table if it exists, making reruns idempotent
- `option("overwriteSchema", "true")` — allows the new write to redefine column types (Delta options accept strings or booleans; the docs use `"true"` as a string)

### D3. Time Travel in Practice

Every write to a Delta table creates a new **version**. `DESCRIBE HISTORY` lists all versions; reads can specify any prior version by number or timestamp. This is the mechanism that makes training data **reproducible** — if you trained a model on Monday's snapshot and want to retrain on exactly the same input two weeks later, you can.

## E. Conclusion

In this lecture, you covered the four platform building blocks that underpin every demo and lab in this module:

1. **Databricks Data + AI Platform** — a unified lakehouse with a Data Intelligence Engine, open formats, and a single governance layer for all ML assets.
2. **Databricks Notebooks** — multi-language, collaborative, and AI-assisted via Genie Code; `display()` and `summary()` are your fastest EDA tools.
3. **Unity Catalog** — a three-level namespace (`catalog.schema.object`) with SQL-based permissions, automated lineage, and governance for data and AI assets such as tables, volumes, functions, and models. Data profiling in Unity Catalog automates data-quality and drift checks on a schedule.
4. **Delta Lake** — the default table format on Databricks; ACID transactions, schema enforcement, and time travel make training data reliable and reproducible. Because it is open, the same tables work across engines, and ingestion options such as Auto Loader, `COPY INTO`, and Lakeflow Connect land data in bronze, silver, and gold layers.
