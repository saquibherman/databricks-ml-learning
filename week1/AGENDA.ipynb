## Data Preparation for Machine Learning

This course teaches the fundamentals of preparing data for traditional machine learning on Databricks. It covers the work that happens before a model is trained, in three modules:

- **Managing and Exploring Data:** load data into Delta tables governed by Unity Catalog, and explore it with notebooks, built-in visualizations, and Genie Code.
- **Data Preparation and Feature Engineering:** clean data, handle missing values and outliers, split and sample it, and build reproducible Spark ML pipelines for imputation, encoding, and standardization.
- **Feature Engineering in Unity Catalog:** store, update, and govern features as feature tables so they can be reused for training and inference, build leakage-free training sets with point-in-time lookups, and see how features are served online through Lakebase-backed Online Feature Stores.

Each module combines lectures with hands-on demos and a lab. Model training, serving, and monitoring are not covered here. They are taught in the *Machine Learning Model Development* and *Machine Learning Model Deployment* courses.

This course is designed for associate-level data scientists and machine learning practitioners who want a solid foundation in data preparation on Databricks.

----

## Course Objectives
By the end of this course, you will be able to:

1. Manage and explore machine learning datasets on Databricks using Unity Catalog governance, Delta Lake time travel, and Genie Code-assisted exploratory data analysis.
1. Diagnose and fix data-quality issues such as incorrect types, invalid values, outliers, and missing data, choosing a strategy that fits the data.
1. Split and sample data for model training and validation, including stratified sampling, time-based and group-based splits, and cross-validation.
1. Engineer features with Spark ML (imputation, encoding, and standardization) and combine them into a reproducible pipeline that is fit on training data only.
1. Create, update, and govern feature tables with Feature Engineering in Unity Catalog, and explain point-in-time lookups and online feature serving.
