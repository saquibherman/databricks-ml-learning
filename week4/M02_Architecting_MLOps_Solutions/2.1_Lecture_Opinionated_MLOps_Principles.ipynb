# Lecture — Opinionated MLOps Principles

## Overview

Before architecting an MLOps solution, it helps to adopt a set of guiding principles. This lecture presents Databricks' opinionated view of how ML operations should be structured — a data-centric foundation, clearly-defined environments, a strategy for separating them, and a pattern for how work moves toward production.

This lecture covers 4 sections that build on each other:

- **A. The Guiding Principle**. A data-centric approach to machine learning — ML projects are made up of data pipelines, aligned to the Data Intelligence Platform vision.
- **B. Multi-Environment Semantics**. Defining the **Development**, **Staging**, and **Production** environments and what happens in each.
- **C. Environment Separation**. Two ways to separate environments — **Direct** (separate workspaces) and **Indirect** (one workspace with enforced separation) — and when to use each.
- **D. Deployment Patterns**. Moving from **Deploy Model** to **Deploy Code** (recommended), and the hybrid option in between.

## Learning Objectives

By the end of this lecture, you will be able to:

1. Explain the data-centric guiding principle behind MLOps on Databricks.
2. Describe the purpose and activities of the Development, Staging, and Production environments.
3. Compare direct and indirect environment separation and choose the right approach.
4. Contrast the Deploy Model and Deploy Code deployment patterns.

## A. The Guiding Principle

### A1. A Data-Centric Approach to ML

Databricks advocates for a **data-centric approach** to machine learning and AI, emphasizing the integration of data pipelines throughout ML projects. This is aligned to the **Data Intelligence Platform vision** — centralized data storage and sharing, with unified security and governance.

- **ML projects are made up of data pipelines.**
- Operationalizing ML solutions requires **connecting a variety of data pipelines** — data preparation, feature engineering, model training, tuning, and inference.
- Data pipelines require **storage, governance, orchestration**, etc.
- Aligned to the **Data Intelligence Platform vision**.

## B. Multi-Environment Semantics

### B1. Development, Staging, and Production

A **multi-environment setup** is a structured approach to ML projects that uses multiple environments — **Development**, **Staging**, and **Production** — to optimize and secure the workflow. Each environment operates on the same three assets: **code**, **data**, and **models**.

## C. Environment Separation

### C1. Direct vs Indirect Separation

A key architectural question is: **how many Databricks workspaces do we have?** There are two ways to separate the Development, Staging, and Production environments.

<div style="display: flex; gap: 16px; flex-wrap: wrap; margin: 16px 0; max-width: 1000px;">
  <div style="flex: 1; min-width: 300px; border: 1px solid #e0e0e0; border-top: 4px solid #FF3621; border-radius: 8px; padding: 16px 20px; background: #FFFFFF;">
    <div style="font-size: 18px; font-weight: 700; color: #1B3139; margin-bottom: 10px;">Direct Separation</div>
    <div style="font-size: 16px; color: #607D8B; margin-bottom: 10px;">Separate workspaces: [ dev ] &rarr; [ staging ] &rarr; [ prod ]</div>
    <ul style="margin: 0; padding-left: 20px; font-size: 16px; line-height: 1.7; color: #333;">
      <li>Completely separate Databricks workspaces for each environment</li>
      <li>Simpler individual environments</li>
      <li>Scales well to multiple projects</li>
      <li><em>Requires more setup and distinct permissions per workspace</em></li>
    </ul>
  </div>
  <div style="flex: 1; min-width: 300px; border: 1px solid #e0e0e0; border-top: 4px solid #FF3621; border-radius: 8px; padding: 16px 20px; background: #FFFFFF;">
    <div style="font-size: 18px; font-weight: 700; color: #1B3139; margin-bottom: 10px;">Indirect Separation</div>
    <div style="font-size: 16px; color: #607D8B; margin-bottom: 10px;">One workspace: [ dev &rarr; staging &rarr; prod ]</div>
    <ul style="margin: 0; padding-left: 20px; font-size: 16px; line-height: 1.7; color: #333;">
      <li>One Databricks workspace with enforced separation (access controls &amp; naming conventions)</li>
      <li>Simpler overall infrastructure, requiring fewer permissions</li>
      <li>More complex individual environment</li>
      <li>Doesn't scale well to multiple projects</li>
    </ul>
  </div>
</div>

<div style="border-left: 4px solid #00A972; background: rgba(0,169,114,0.10); padding: 14px 18px; border-radius: 4px; margin: 16px 0; font-family: sans-serif;">
  <strong style="display:block; color:#00A972; margin-bottom:6px; font-size:15pt;">Success</strong>
  <div style="color:#0b2026; font-size:12pt; line-height:1.55;">
    Adopting the right environment separation strategy is a best practice for robust MLOps. 
    <br><br>
    <strong>Direct Separation</strong> is ideal for larger teams and multiple projects, ensuring strong isolation and scalability. 
    <br>
    <strong>Indirect Separation</strong> simplifies infrastructure for smaller teams or projects, making it easier to learn and manage.
    <br><br>
    <em>This course uses Indirect Separation to help you focus on core MLOps principles without extra complexity.</em>
  </div>
</div>

## D. Deployment Patterns

### D1. Deploy Model vs Deploy Code

There are two deployment patterns for moving an ML solution toward production: **Deploy Model** and **Deploy Code** (recommended). The difference is *what* moves through the environments — a trained model artifact, or the training code itself.

<div style="display: flex; gap: 16px; flex-wrap: wrap; margin: 16px 0; max-width: 1000px;">
  <div style="flex: 1; min-width: 300px; border: 1px solid #e0e0e0; border-top: 4px solid #1B5162; border-radius: 8px; padding: 16px 20px; background: #FFFFFF;">
    <div style="font-size: 18px; font-weight: 700; color: #1B3139; margin-bottom: 10px;">Deploy Model</div>
    <ul style="margin: 0; padding-left: 20px; font-size: 16px; line-height: 1.7; color: #333;">
      <li>Model is trained in the <strong>development</strong> environment</li>
      <li>Model artifact is moved from <strong>staging</strong> through <strong>production</strong></li>
      <li>A separate process is needed for other code (inference, monitoring, operational pipelines)</li>
    </ul>
    <p style="font-size: 16px; color: #607D8B; margin: 12px 0 0 0;"><em>Suitable when model training is expensive or hard to reproduce.</em></p>
  </div>
  <div style="flex: 1; min-width: 300px; border: 1px solid #e0e0e0; border-top: 4px solid #00A972; border-radius: 8px; padding: 16px 20px; background: #FFFFFF;">
    <div style="font-size: 18px; font-weight: 700; color: #1B3139; margin-bottom: 10px;">Deploy Code <span style="color:#00A972;">(recommended)</span></div>
    <ul style="margin: 0; padding-left: 20px; font-size: 16px; line-height: 1.7; color: #333;">
      <li>Code is developed in the <strong>development</strong> environment</li>
      <li>Code is tested in the <strong>staging</strong> environment</li>
      <li>Code is deployed in the <strong>production</strong> environment</li>
      <li>The training pipeline is run in each environment; the model is deployed in production</li>
    </ul>
    <p style="font-size: 16px; color: #607D8B; margin: 12px 0 0 0;"><em>Databricks recommends this for most situations.</em></p>
  </div>
</div>

With **Deploy Code**, the model is trained in each environment: starting in development for model development, then staging for integration tests on a limited subset of data, and finally in production on the full production data to produce the final model.

<div style="border-left: 4px solid #00A972; background: rgba(0,169,114,0.10); padding: 14px 18px; border-radius: 4px; margin: 16px 0; font-family: sans-serif;">
  <strong style="display:block; color:#00A972; margin-bottom:6px; font-size:15pt;">Success</strong>
  <div style="color:#0b2026; font-size:12pt; line-height:1.55;"><strong>Benefits of Deploy Code:</strong> Useful when access to production data is restricted, and enables <strong>safer automated model retraining</strong> — the training code is reviewed, tested, and approved for production. The main disadvantage is a steeper learning curve for data scientists to hand off code to collaborators.</div>
</div>

<div style="border-left: 4px solid #4299E0; background: rgba(66,153,224,0.10); padding: 14px 18px; border-radius: 4px; margin: 16px 0; font-family: sans-serif;">
  <strong style="display:block; color:#2272B4; margin-bottom:6px; font-size:15pt;">Info</strong>
  <div style="color:#0b2026; font-size:12pt; line-height:1.55;"><strong>Hybrid approach:</strong> When a situation requires training the model in staging over the full production dataset, code is deployed to staging, the model is trained there, and then the trained model is deployed to production.</div>
</div>

## Conclusion

In this lecture, you covered the opinionated principles behind MLOps on Databricks:

1. **The Guiding Principle** — Databricks takes a **data-centric approach**: ML projects are made up of data pipelines that require storage, governance, and orchestration, all aligned to the Data Intelligence Platform vision.
2. **Multi-Environment Semantics** — ML work is structured across three environments: **Development** (explore, experiment, develop), **Staging** (test and validate), and **Production** (deploy and monitor).
3. **Environment Separation** — Environments can be separated **directly** (separate workspaces — scales to many projects) or **indirectly** (one workspace with enforced separation — simpler). This course uses indirect separation.
4. **Deployment Patterns** — **Deploy Model** moves a trained artifact through environments, while **Deploy Code** (recommended) moves the training code and trains the model in each environment — enabling safer, automated retraining.
