# Lecture — Recommended MLOps Architectures

## Overview

Building on the opinionated MLOps principles — a data-centric approach, multi-environment semantics, and the Deploy Code pattern — this lecture assembles them into a concrete, recommended architecture. It explains why architecture matters, separates the two dimensions of architecture, walks through the four phases of an MLOps solution, and finishes with the complete end-to-end picture.

This lecture covers 4 sections that build on each other:

- **A. Why Architecture Matters**. The benefits of getting architecture right, and the two dimensions of architecture — **Infrastructure** (set up once) vs **Workflow** (ongoing processes).
- **B. The Recommended Architecture**. A high-level view of code, data, and ML environments, and the four phases of an MLOps solution.
- **C. The Phases in Detail**. A closer look at each phase — **Infrastructure Setup**, **Development**, **Staging**, and **Production** — using the DataOps + DevOps + ModelOps mental model.
- **D. The Complete Architecture**. How the phases combine into a single, end-to-end MLOps architecture on Databricks.

## Learning Objectives

By the end of this lecture, you will be able to:

1. Explain why a well-designed MLOps architecture matters and distinguish infrastructure from workflow.
2. Describe the recommended MLOps architecture and the four phases of an MLOps solution.
3. Summarize the key tasks of the Infrastructure, Development, Staging, and Production phases.
4. Describe how the phases combine into a complete MLOps architecture on Databricks.

## Conclusion

In this lecture, you assembled the opinionated principles into a recommended MLOps architecture:

1. **Why Architecture Matters** — A well-designed architecture delivers **simplicity, efficiency, scalability, and collaboration**. Architecture has two dimensions: **Infrastructure** (set up once) and **Workflow** (repeatable, ongoing processes).
2. **The Recommended Architecture** — A **single code repository** and a **single data/artifact management solution** bracket the Development, Staging, and Production environments. An MLOps solution runs in four phases: a one-time **Infrastructure Setup**, then a cyclical loop of **Development → Staging → Production**.
3. **The Phases in Detail** — **Infrastructure** (Architect/Engineer, set up once with Unity Catalog, workspaces, catalogs, MLOps Stacks); **Development** (Data Scientist builds ML pipelines, commits code); **Staging** (ML Engineer runs automated tests via pull request); **Production** (ML Engineer deploys from main, compares champion vs. challenger, monitors for drift).
4. **The Complete Architecture** — All phases combine into one end-to-end architecture on Databricks: a shared Git repository, three MLflow-tracked environments, and a unified Unity Catalog / Lakehouse layer — delivering holistic management, scalability, and collaboration.
