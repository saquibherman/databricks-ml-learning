# Lecture — Monitoring in Machine Learning

## Overview

The previous lecture explained *what* drift is and *why* to watch for it. This lecture shows *how* to do it on Databricks — the practice of **monitoring ML systems**, the **data profiling** capabilities that make monitoring a built-in, managed feature of the platform, and the **Databricks SQL alerts** that turn a detected problem into an automatic notification or retraining trigger.

This lecture covers 4 sections that build on each other:

- **A. Monitoring ML Systems**. What monitoring is, why it matters, and the data and ML assets worth monitoring.
- **B. Data Profiling in Practice**. How data profiling integrates monitoring across DataOps, DevOps, and ModelOps — including inference tables and ground-truth integration.
- **C. Data Profiling Capabilities**. The metrics, time granularities, data slices, and asset types you can monitor, plus dashboards, alerts, and open results.
- **D. Databricks SQL Alerts**. How scheduled queries evaluate conditions and send multi-channel notifications — and how they close the loop back to retraining.

## Learning Objectives

By the end of this lecture, you will be able to:

1. Explain what monitoring an ML system involves and which data and assets to monitor.
2. Describe how data profiling integrates monitoring across the components of MLOps.
3. List the core data profiling capabilities and how they are configured.
4. Describe Databricks SQL alerts and how they trigger notifications and retraining.

## A. Monitoring ML Systems

**Monitoring an ML system** is the *continuous logging and review* of key component and system metrics — the ongoing evaluation used to analyze the performance of a solution in production.

## B. Data Profiling in Practice

### B1. Data Profiling on Databricks

**Data profiling** is Databricks' **centralized, standardized mechanism for monitoring models in production** — a simplified, built-in tool that lets you manage, govern, evaluate, and switch models easily.

<div style="max-width: 1000px; margin: 16px 0;">
  <ul style="margin: 0; padding-left: 20px; font-size: 16px; line-height: 1.8; color: #333;">
    <li><strong>Monitor data and AI assets</strong> from one place</li>
    <li><strong>Centralized and standardized</strong> mechanism for monitoring models in production</li>
    <li><strong>Simplified, built-in tool</strong> for monitoring mechanisms — diagnose errors, detect drift, etc.</li>
    <li>Allows for the creation of <strong>additional custom metrics</strong></li>
    <li><strong>Alerting</strong> to get notified on drift or quality issues</li>
  </ul>
</div>

<div style="border-left: 4px solid #1B5162; background: #F8F9FC; padding: 14px 18px; border-radius: 4px; margin: 16px 0; font-family: sans-serif;">
  <strong style="display:block; color:#1B5162; margin-bottom:6px; font-size:15pt;">Note</strong>
  <div style="color:#0b2026; font-size:12pt; line-height:1.55;">Data profiling is the monitoring capability within <strong>Unity Catalog</strong> (evolved from what earlier docs called <em>Lakehouse Monitoring</em>). Because it is built into the platform, there is no separate infrastructure to stand up — you point it at a table, view, or model and it produces the profile and drift metrics for you.</div>

## C. Data Profiling Capabilities

### C1. Metrics, Granularities, Slices &amp; Assets

Data profiling can be configured from the **UI (Unity Catalog Explorer)** or the **Python API**. Its core capabilities cover *what* metrics to compute, over *what* time windows, on *which* slices, and across *which* asset types.

The Python API consists of essentially **two calls** — an idempotent call to set up (or update) monitoring on a table, and a call to recompute the metrics from the latest contents of the table:

<div style="background: #FFF3E0; border-left: 4px solid #F57C00; border-radius: 6px; padding: 16px; margin: 12px 0; overflow-x: auto;">
<pre style="margin: 0; font-family: 'SFMono-Regular', Consolas, 'Liberation Mono', Menlo, monospace; font-size: 13px; color: #0a0a0a; white-space: pre-wrap;">
<code>import databricks.lakehouse_monitoring as lm
<strong># Set up monitoring parameters (idempotent — the config for monitoring)</strong>
lm.create_monitor(
    table_name="my_catalog.my_schema.my_table",
    # ...
)
<strong># Refresh monitoring metrics based on the latest contents of the table</strong>
lm.run_refresh("my_catalog.my_schema.my_table")
</code>
</pre>
</div>

## D. Databricks SQL Alerts

### D1. What Are Databricks SQL Alerts?

**Databricks SQL alerts** periodically run queries, evaluate defined conditions, and send notifications if a condition is met. They are how a *detected* drift or quality issue becomes an *action*.

<div style="display: flex; gap: 16px; flex-wrap: wrap; margin: 16px 0; max-width: 1000px;">
  <div style="flex: 1; min-width: 300px; border: 1px solid #e0e0e0; border-top: 4px solid #FF3621; border-radius: 8px; padding: 16px 20px; background: #FFFFFF;">
    <div style="font-size: 17px; font-weight: 700; color: #1B3139; margin-bottom: 8px;">Scheduled Execution</div>
    <div style="font-size: 15px; color: #333; line-height: 1.6;">Automatically runs queries at defined intervals to check specific conditions. Each alert runs <strong>independently</strong> on its own schedule.</div>
  </div>
  <div style="flex: 1; min-width: 300px; border: 1px solid #e0e0e0; border-top: 4px solid #00A972; border-radius: 8px; padding: 16px 20px; background: #FFFFFF;">
    <div style="font-size: 17px; font-weight: 700; color: #1B3139; margin-bottom: 8px;">Multi-Channel Notifications</div>
    <div style="font-size: 15px; color: #333; line-height: 1.6;">Receive alerts via <strong>Email, Slack, Webhook, MS Teams, PagerDuty</strong>, and more — facilitating immediate action and response.</div>
  </div>
</div>

A SQL alert evaluates a **trigger condition** — an aggregate of a value column against an operator and threshold (for example, `MAX(wasserstein_distance) > 0.15`) — and fires when the condition is met. Pointed at a monitor's drift metrics table, it turns a drift measurement into an action:

<div style="background: #FFF3E0; border-left: 4px solid #F57C00; border-radius: 6px; padding: 16px; margin: 12px 0; overflow-x: auto;">
<pre style="margin: 0; font-family: 'SFMono-Regular', Consolas, 'Liberation Mono', Menlo, monospace; font-size: 13px; color: #0a0a0a; white-space: pre-wrap;">
<code>-- The query an alert runs on a schedule; the alert fires when the condition is met</code>
<code>SELECT MAX(wasserstein_distance) AS max_drift</code>
<code>FROM my_catalog.my_schema.my_table_drift_metrics</code>
<code>WHERE column_name = 'credit_score'</code>
<code>  AND window_end &gt;= current_date() - INTERVAL 1 DAY;</code>
<code>-- Alert condition:  max_drift  &gt;  0.15</code>
</pre>
</div>

<div style="border-left: 4px solid #00A972; background: rgba(0,169,114,0.10); padding: 14px 18px; border-radius: 4px; margin: 16px 0; font-family: sans-serif;">
  <strong style="display:block; color:#00A972; margin-bottom:6px; font-size:15pt;">Success</strong>
  <div style="color:#0b2026; font-size:12pt; line-height:1.55;">Set alert schedules <strong>tighter than your drift-detection window</strong> so an issue is caught within one refresh cycle, and point the alert query at the monitor's <strong>drift metrics table</strong> rather than a raw business table — so you are alerting on drift, not on normal fluctuations in the underlying data.</div>

## Conclusion

In this lecture, you covered how to monitor machine learning on Databricks:

1. **Monitoring ML Systems** — Monitoring is the continuous logging and review of key metrics to diagnose issues before they become severe or costly. You monitor **data** (input, feature stores, vector databases, human feedback, model outputs) and **ML assets** (checkpoints, component and system evaluation metrics, performance/cost).
2. **Data Profiling in Practice** — Data profiling is Databricks' centralized, built-in mechanism for monitoring models in production, integrating across DataOps, DevOps, and ModelOps. **Inference tables** log every prediction and its output, and logging **ground truth** alongside enables continuous validation.
3. **Data Profiling Capabilities** — Configure via UI or a two-call Python API. Monitor key metrics, define multiple time granularities, slice by columns or predicates, and cover tables, views, and ML models — with auto-generated dashboards, alerts, open Delta results, and managed compute.
4. **Databricks SQL Alerts** — Scheduled queries evaluate conditions and send multi-channel notifications, closing the loop by triggering retraining, Lakeflow Jobs, or CI/CD pipelines.
