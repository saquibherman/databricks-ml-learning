# Lecture — Evaluating Model Performance

## Overview

This lecture covers how to measure and compare machine learning model quality. You will learn the end-to-end model development workflow, how to structure data splits for reliable assessment, and which metrics to use for regression, classification, and clustering tasks.

This lecture covers 4 sections that build on each other:

- **A. Purpose of Evaluation Metrics**. Why we compute metrics, the end-to-end model development workflow (prepare → split → train → tune → evaluate), and the roles of train/validation/test splits. Tools: MLflow for experiment tracking, Optuna for hyperparameter tuning.
- **B. Regression Metrics**. R², Mean Absolute Error, Mean Squared Error, and Root Mean Squared Error — when to use each and how to compute them.
- **C. Classification Metrics**. The confusion matrix, accuracy, precision, recall, F1 score, log loss (cross-entropy), and ROC/AUC — choosing the right metric based on the business cost of errors.
- **D. Clustering Metrics**. K-Means fundamentals, the silhouette score for measuring cluster separation, and the elbow method for choosing the optimal number of clusters.

## Learning Objectives

By the end of this lecture, you will be able to:

1. Describe the end-to-end model development workflow and explain the distinct role of training, validation, and test data in model selection.
2. Select and interpret the appropriate metrics for regression, classification, and clustering tasks.
3. Choose between precision, recall, and F1 based on the business cost of different error types.

## B. Regression Metrics

### B1. Common Regression Metrics

Regression models predict continuous outputs. These four metrics quantify prediction error in different ways.

<table style="border-collapse: collapse; width: 100%; max-width: 860px; font-family: -apple-system, BlinkMacSystemFont, 'Segoe UI', Roboto, sans-serif; font-size: 14px; margin: 16px 0;">
  <thead>
    <tr style="background-color: #1B3139;">
      <th style="padding: 12px 16px; text-align: left; color: #ffffff; border-bottom: 3px solid #FF3621;">Metric</th>
      <th style="padding: 12px 16px; text-align: left; color: #ffffff; border-bottom: 3px solid #FF3621;">Description</th>
      <th style="padding: 12px 16px; text-align: left; color: #ffffff; border-bottom: 3px solid #FF3621;">Direction</th>
    </tr>
  </thead>
  <tbody>
    <tr style="background-color: #F9F7F4;">
      <td style="padding: 12px 16px; border-bottom: 1px solid #e0e0e0; font-weight: 600;">R² (Coefficient of Determination)</td>
      <td style="padding: 12px 16px; border-bottom: 1px solid #e0e0e0;">Measures how well the model explains the variation in the target variable. At most 1 (a perfect fit); can be negative when the model is worse than predicting the mean.</td>
      <td style="padding: 12px 16px; border-bottom: 1px solid #e0e0e0;">Larger is better</td>
    </tr>
    <tr>
      <td style="padding: 12px 16px; border-bottom: 1px solid #e0e0e0; font-weight: 600;">Mean Absolute Error (MAE)</td>
      <td style="padding: 12px 16px; border-bottom: 1px solid #e0e0e0;">Average absolute difference between actual and predicted values.</td>
      <td style="padding: 12px 16px; border-bottom: 1px solid #e0e0e0;">Lower is better</td>
    </tr>
    <tr style="background-color: #F9F7F4;">
      <td style="padding: 12px 16px; border-bottom: 1px solid #e0e0e0; font-weight: 600;">Mean Squared Error (MSE)</td>
      <td style="padding: 12px 16px; border-bottom: 1px solid #e0e0e0;">Average of squared differences between actual and predicted values. Squaring gives higher weight to large errors.</td>
      <td style="padding: 12px 16px; border-bottom: 1px solid #e0e0e0;">Lower is better</td>
    </tr>
    <tr>
      <td style="padding: 12px 16px; border-bottom: 1px solid #e0e0e0; font-weight: 600;">Root Mean Squared Error (RMSE)</td>
      <td style="padding: 12px 16px; border-bottom: 1px solid #e0e0e0;">Square root of MSE, making it interpretable in the original units of the target.</td>
      <td style="padding: 12px 16px; border-bottom: 1px solid #e0e0e0;">Lower is better</td>
    </tr>
  </tbody>
</table>

What are the four cells of the confusion matrix?

<div style="border: 2px solid #1B5162; border-top: none; border-radius: 0 0 8px 8px; padding: 16px 20px; background: #F9F7F4; font-size: 13pt; line-height: 1.7; color: #333;">

<ul style="margin: 12px 0; padding-left: 24px;">
  <li><strong>True Positive (TP)</strong> — predicted positive, actual positive. Correctly identified what we were looking for.</li>
  <li><strong>False Negative (FN)</strong> — predicted negative, actual positive. Missed it. Can be very costly: missed disease, missed fraud, missed security breach.</li>
  <li><strong>False Positive (FP)</strong> — predicted positive, actual negative. False alarm. Creates noise, wasted effort, and sometimes harm.</li>
  <li><strong>True Negative (TN)</strong> — predicted negative, actual negative. Correctly said "nothing here." Often the largest cell on imbalanced data, which is why raw accuracy can be misleading.</li>
</ul>
<p>The confusion matrix is the single most useful visualization for any classifier — from these four counts you derive every classification metric: accuracy, precision, recall, F1, and so on.</p>

These four metrics are derived from the confusion matrix, each emphasizing different aspects of classification quality.

      <td style="padding: 12px 16px; border-bottom: 1px solid #e0e0e0;">Proportion of correctly classified instances. Can be misleading for imbalanced datasets.</td>
      <td style="padding: 12px 16px; border-bottom: 1px solid #e0e0e0;">Balanced class distributions</td>

    <tr>
      <td style="padding: 12px 16px; border-bottom: 1px solid #e0e0e0; font-weight: 600;">Precision</td>

    </tr>
    <tr>
      <td style="padding: 12px 16px; border-bottom: 1px solid #e0e0e0; font-weight: 600;">F1 Score</td>
      <td style="padding: 12px 16px; border-bottom: 1px solid #e0e0e0;">Harmonic mean of precision and recall. Balances both concerns in a single number.</td>
      <td style="padding: 12px 16px; border-bottom: 1px solid #e0e0e0;">Need balance between precision and recall</td>
    </tr>

### C3. Choosing Between Precision, Recall, and F1

From the confusion matrix come the three metrics you'll use most often in classification. The right one to optimize depends on the business cost: high precision matters when false positives are expensive, high recall when false negatives are costly, and F1 when you need a balance between the two.

### D2. The Elbow Method

K-Means is the classic clustering algorithm. You pick K, the number of clusters, and the algorithm iteratively assigns points to the nearest cluster centroid and then recomputes centroids until they stop moving. Simple, fast, and surprisingly effective — but you have to choose K, and the algorithm assumes clusters are roughly spherical and similar in size. Those assumptions don't always hold, which is why alternatives such as DBSCAN or Gaussian mixtures — which relax them — exist.

## E. Conclusion

In this lecture, you covered the end-to-end model development workflow and how to measure model quality across ML paradigms:

1. **Purpose of Evaluation Metrics & the Model Development Workflow** — Metrics quantify performance, enable comparison, guide tuning, and provide objective model selection. The workflow runs from data preparation through split strategy, training (tracked with MLflow), hyperparameter tuning (Optuna), and final evaluation on the held-out test set. Training data directly updates parameters; validation data indirectly influences them; test data has no influence — it only measures generalizability.
2. **Regression Metrics** — R² measures explained variance; MAE, MSE, and RMSE quantify prediction error at different scales. Use `root_mean_squared_error` for interpretable error in original units.
3. **Classification Metrics** — The confusion matrix is the foundation — from its four cells you derive accuracy, precision, recall, and F1. Choose based on business cost: precision when false positives are expensive, recall when false negatives are. Log loss measures probability calibration (confidence matters, not just correctness). ROC/AUC provides threshold-independent model comparison.
4. **Clustering Metrics** — Without labels, use the silhouette score to measure cluster separation. K-Means is the classic algorithm, but you must choose K — the elbow method helps by plotting inertia against cluster count and finding diminishing returns. Domain knowledge plays a role; alternatives like DBSCAN relax K-Means' spherical-cluster assumption.
