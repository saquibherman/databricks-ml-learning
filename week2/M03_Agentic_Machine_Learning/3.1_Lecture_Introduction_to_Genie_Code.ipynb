# Lecture — Introduction to Genie Code

## Overview

This lecture introduces Genie Code — Databricks' AI coding and data assistant for developers and technical practitioners — and how it supports machine learning model development. You will learn how Genie Code addresses the challenges of traditional ML workflows, explore its key capabilities, understand platform integrations, and discover how to customize it with MCP connections, instructions, and skills.

This lecture covers 4 sections that build on each other:

- **A. The ML Development Challenge**. Why building ML models is complex, time-consuming work — and how Genie Code provides a natural-language solution that accelerates the entire process.
- **B. What Is Genie Code**. The core identity, design principles, and five key ML needs that Genie Code addresses — spanning ML models, pipelines, and dashboards.
- **C. Platform Integration and Customization**. How Genie Code integrates with Databricks services and how you customize it with MCP, instructions, and skills.
- **D. End-to-End Model Development**. A complete walkthrough from data exploration to deployed endpoint — all in one conversation.

## Learning Objectives

By the end of this lecture, you will be able to:

- Describe the challenges of traditional ML development workflows and how Genie Code addresses them.
- Describe the key ML capabilities of Genie Code on the Databricks platform.
- Customize Genie Code with MCP connections, instructions, and skills.
- Use Genie Code to perform end-to-end model development through natural language.

## C. Platform Integration and Customization

### C1. Genie Code Is AI That Knows the Platform

Genie Code understands and integrates with other Databricks products.

**Genie Code integrates with:**

- **Feature Store**: Expand the input training dataset using existing feature tables.
- **MLflow Model Tracking**: All trial run metrics and parameters are tracked.
- **Model Registry**: Register generated models to Unity Catalog.
- **Model Serving**: Stand up low-latency model-serving endpoints with the right specs — Genie Code handles endpoint configuration for real-time inference.
- **Lakeflow Pipelines**: Create and maintain Spark Declarative Pipelines.
- **Multi-Language Notebooks**: Works across Python, SQL, and Scala — no context-switching as you move from data prep to modeling to evaluation.
- **And more...** including Data Versioning, Runtime Environments, Batch Scoring, Online Serving, and Monitoring — all governed through Unity Catalog.

### C4. Guide Genie Code with Instructions

Define clear guidance and structure to improve response quality.

**Instructions:**

- Markdown files that give Genie Code **always-on context**: your team's coding standards, preferred libraries, naming conventions, and the language you want responses in.
- **Configurable** at workspace or user level — scope them broadly to apply everywhere, or narrowly for personal preferences.

<details style="margin: 16px 0;">
<summary style="background: linear-gradient(135deg, #1B5162, #4299E0); color: white; padding: 12px 18px; cursor: pointer; font-weight: 600; font-size: 13pt; border-radius: 8px; user-select: none;">
Best practices for Instructions
</summary>
<div style="border: 2px solid #1B5162; border-top: none; border-radius: 0 0 8px 8px; padding: 16px 20px; background: #F9F7F4; font-size: 13pt; line-height: 1.7; color: #333;">

<ul style="margin: 12px 0; padding-left: 24px;">

</ul>

### C5. Equip Genie Code with Skills

Operationalize domain knowledge through reusable workflows and automation.

**Skills:**

- On-demand **bundles of domain knowledge** and workflows — markdown plus optional scripts — that Genie Code pulls in when invoked in Agent mode.
- Think of skills as **recipes for common tasks**: a *"deploy this model"* skill, a *"build a churn model"* skill, a *"set up monitoring"* skill.
- Build them once and reuse them across the team.

<details style="margin: 16px 0;">
<summary style="background: linear-gradient(135deg, #1B5162, #4299E0); color: white; padding: 12px 18px; cursor: pointer; font-weight: 600; font-size: 13pt; border-radius: 8px; user-select: none;">
