## A. Review of Regression Metrics

### A1. Regression Metrics

Regression models predict continuous outputs. These four metrics quantify prediction error in different ways.

      <td style="padding: 12px 16px; border-bottom: 1px solid #e0e0e0;">2&times;(0.75&times;0.60)/(0.75+0.60) = <strong>2/3</strong></td>

## Conclusion

This appendix reviewed the foundational evaluation metrics for supervised learning:

**Regression Metrics:**
* The **residual** is the core building block — the distance between each observed value and the model's prediction
* **R²** measures how much of the total variability (SS_tot) the model explains by comparing model error (SS_res) to a baseline mean prediction
* **MAE** averages the absolute residuals — interpretable and robust to outliers
* **MSE** averages the squared residuals — penalizes large errors more heavily and is differentiable for optimization
* **RMSE** is the square root of MSE — returns the error to the original units of the target variable

**Classification Metrics:**
* The **confusion matrix** (TP, FP, FN, TN) is the foundation for all classification metrics
* **Precision** and **Recall** capture different error costs — choose based on whether false positives or false negatives are more harmful
* **F1 Score** balances precision and recall into a single number
* **Log Loss** penalizes confident wrong predictions steeply — it evaluates predicted probabilities, not just class labels

**Bias-Variance Tradeoff:**
* **High bias** (underfitting) means the model is too simple to capture patterns — both training and validation errors are high
* **High variance** (overfitting) means the model memorizes noise — training error is low but validation error is high
* The goal is to find the **sweet spot** where the gap between training and validation error is minimized
